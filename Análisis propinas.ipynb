```python
# Análisis exploratorio

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set(style="darkgrid")

df = sns.load_dataset("tips")
df.head()
```

```python
# Gráfico de barras de propinas por sexo

sns.barplot(data=df, x="sex", y="tip", hue="sex", errorbar=None)
plt.title("Propinas promedio por sexo")
plt.figtext(0.5, -0.02, "Análisis general: En promedio los hombres dan más propinas", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Gráfico de barras de propinas por día

sns.barplot(data=df, x="day", y="tip", hue="day", errorbar=None)
plt.title("Propinas promedio por día")
plt.figtext(0.5, -0.02, "Análisis general: En promedio los domingos se dan más propinas", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Gráfico de barras de propinas por tiempo

sns.barplot(data=df, x="time", y="tip", hue="time", errorbar=None)
plt.title("Propinas promedio por tiempo")
plt.figtext(0.5, -0.02, "Análisis general: En promedio durante las cenas se dan más propinas", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Gráfico de barras de propinas por fumador/no fumador

sns.barplot(data=df, x="smoker", y="tip", hue="smoker", errorbar=None)
plt.title("Propinas promedio por fumador/no fumador")
plt.figtext(0.5, -0.02, "Análisis general: El hecho de que un cliente fume o no, no afecta a la propina", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Gráfico de barras de propinas (días/sexo)

sns.barplot(data=df, x="day", y="tip", hue="sex", errorbar=None)
plt.title("Propinas promedio por día y sexo")
plt.figtext(0.5, -0.02, "Análisis general: En promedio los hombres en días domingos dan más propinas, que las mujeres", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Gráfico de barras de propinas (tiempo/sexo)

sns.barplot(data=df, x="time", y="tip", hue="sex", errorbar=None)
plt.title("Propinas promedio por día y sexo")
plt.figtext(0.5, -0.02, "Análisis general: En promedio durante las cenas los hombres dan más propinas, en comparación a las mujeres", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Gráfico de barras de propinas (fumador/sexo)

sns.barplot(data=df, x="smoker", y="tip", hue="sex", errorbar=None)
plt.title("Propinas promedio por fumador/no fumador y sexo")
plt.figtext(0.5, -0.02, "Análisis general: Los hombres (fumadores y no fumadores) son más propensos a dar propina, con respecto a las mujeres (fumadoras y no fumadoras)", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Gráfico de dispersión de propinas (días/sexo)

sns.scatterplot(data=df, x="day", y="tip")
plt.title("Relación entre el tamaño de la fiesta y propinas")
plt.figtext(0.5, -0.02, "Análisis general: Cuántas más personas hayan en una fiesta existe mayor posibilidad de que den propinas", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Gráfico de barras de propinas por tamaño de la fiesta

sns.barplot(data=df, x="size", y="tip", hue="size", errorbar=None)
plt.title("Propinas promedio por tamaño de la fiesta")
plt.figtext(0.5, -0.02, "Análisis general: Cuántas más personas hayan en una fiesta existe mayor posibilidad de que den propinas", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Cálculo del porcentaje de propina

df["tip_percentage"] = (df["tip"] / df["total_bill"]) * 100
df.head()
```

```python
# Gráfico de barras de porcentaje de propinas por sexo

sns.barplot(data=df, x="sex", y="tip_percentage", hue="sex", errorbar=None)
plt.title("Porcentaje de propinas promedio por sexo")
plt.figtext(0.5, -0.02, "Análisis general: En términos porcentuales, las mujeres dan mayores propinas, que los hombres", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Gráfico de barras de porcentaje de propinas por día

sns.barplot(data=df, x="day", y="tip_percentage", hue="day", errorbar=None)
plt.title("Porcentaje de propinas promedio por día")
plt.figtext(0.5, -0.02, "Análisis general: En términos porcentuales, los viernes se dan mayores propinas", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Gráfico de barras de porcentaje de propinas por tiempo

sns.barplot(data=df, x="time", y="tip_percentage", hue="time", errorbar=None)
plt.title("Porcentaje de propinas promedio por tiempo")
plt.figtext(0.5, -0.02, "Análisis general: En términos porcentuales, en los desayunos se dan mayores propinas", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```

```python
# Gráfico de barras de porcentaje de propinas por fumador/no fumador

sns.barplot(data=df, x="smoker", y="tip_percentage", hue="smoker", errorbar=None)
plt.title("Porcentaje de propinas promedio por fumador/no fumador")
plt.figtext(0.5, -0.02, "Análisis general: En términos porcentuales, ligeremente los clientes fumadores dan mayores propinas que los no fumadores", 
            ha="center", fontsize=9, style="italic", color="gray")
plt.show()
```
